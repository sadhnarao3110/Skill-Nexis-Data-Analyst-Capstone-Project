```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
df.columns = df.columns.str.strip()
print(df.columns.tolist())
```

```python
df.loc[df["Discount Band"].isnull()].head(10)
```

```python
df.loc[
    df["Discount Band"].isnull(),
    ["Gross Sales", "Discounts", "Sales", "Discount Band"]
].head(10)
```

```python
df["Discount Band"] = df["Discount Band"].fillna("Unknown")

print("Missing values after handling:")
print(df.isnull().sum())
```

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    errors="coerce",
    dayfirst=True
)

print(df["Date"].dtype)
df[["Date"]].head()
```

```python
print("Invalid or missing dates:", df["Date"].isna().sum())
```

```python
numeric_columns = [
    "Units Sold",
    "Manufacturing Price",
    "Sale Price",
    "Gross Sales",
    "Discounts",
    "Sales",
    "COGS",
    "Profit",
    "Month Number",
    "Year"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df[numeric_columns].dtypes)
```

```python
# Profit margin as a percentage of sales
df["Profit Margin %"] = np.where(
    df["Sales"] != 0,
    (df["Profit"] / df["Sales"]) * 100,
    0
)

# Discount rate as a percentage of gross sales
df["Discount Rate %"] = np.where(
    df["Gross Sales"] != 0,
    (df["Discounts"] / df["Gross Sales"]) * 100,
    0
)

# Quarter derived from the date
df["Quarter"] = "Q" + df["Date"].dt.quarter.astype("Int64").astype(str)

df[["Sales", "Profit", "Profit Margin %", "Discount Rate %", "Quarter"]].head()
```

```python
print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nFinal dataset preview:")
display(df.head())
```

```python
df.to_csv("financial_data_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
df = pd.read_csv("financial_data_cleaned.csv")
```

```python
df = df.dropna(subset=["Sales"]).copy()

features = [
    "Units Sold",
    "Manufacturing Price",
    "Sale Price",
    "Discounts",
    "Month Number",
    "Year",
    "Segment",
    "Country",
    "Product",
    "Discount Band"
]

features = [col for col in features if col in df.columns]

X = df[features]
y = df["Sales"]

print("Input features:", features)
print("Target variable: Sales")
print("Number of records:", len(df))
```

```python
from sklearn.model_selection import train_test_split

print("train_test_split imported successfully!")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

print("All preprocessing libraries imported successfully!")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("All libraries imported successfully!")
```

```python
numeric_features = X.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(transformers=[
    ("numeric", numeric_transformer, numeric_features),
    ("categorical", categorical_transformer, categorical_features)
])

print("Numeric features:", numeric_features)
print("Categorical features:", categorical_features)
```

```python
model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", LinearRegression())
])

model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
```

```python
y_pred = model.predict(X_test)

results = pd.DataFrame({
    "Actual Sales": y_test.values,
    "Predicted Sales": y_pred
})

results["Prediction Error"] = (
    results["Actual Sales"] - results["Predicted Sales"]
)

display(results.head(10))
```

```python
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"Mean Absolute Error (MAE): {mae:,.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:,.2f}")
print(f"R-squared (R²): {r2:.4f}")
print(f"R-squared (%): {r2 * 100:.2f}%")
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred, alpha=0.7)

min_value = min(y_test.min(), y_pred.min())
max_value = max(y_test.max(), y_pred.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.xlabel("Actual Sales")
plt.ylabel("Predicted Sales")
plt.title("Actual vs Predicted Sales - Linear Regression")
plt.tight_layout()
plt.show()
```

```python
results.to_csv("sales_regression_predictions.csv", index=False)

metrics = pd.DataFrame({
    "Metric": ["MAE", "RMSE", "R2"],
    "Value": [mae, rmse, r2]
})

metrics.to_csv("sales_regression_metrics.csv", index=False)

print("Predictions and evaluation metrics saved successfully!")
```
